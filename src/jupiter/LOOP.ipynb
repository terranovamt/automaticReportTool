```python
def ordina_test_name(test_names):
    def sort_key(name):
        prefix, suffix = name.split(":", 1) if ":" in name else (name, "")
        if "Untrimmed" in suffix:
            return (prefix, 0)
        elif "TrimValue" in suffix:
            return (prefix, 1)
        elif "Trimmed" in suffix:
            return (prefix, 2)
        return (
            prefix,
            3,
        )  # Per tutti gli altri nomi che non contengono le parole chiave

    # Mantieni l'ordine generale dei test ma ordina all'interno dei gruppi
    grouped_tests = {}
    for test in test_names:
        prefix = test.split(":", 1)[0]
        if prefix not in grouped_tests:
            grouped_tests[prefix] = []
        grouped_tests[prefix].append(test)

    sorted_tests = []
    for prefix in test_names:
        prefix_key = prefix.split(":", 1)[0]
        if prefix_key in grouped_tests:
            sorted_tests.extend(sorted(grouped_tests[prefix_key], key=sort_key))
            del grouped_tests[prefix_key]  # Rimuovi il gruppo una volta aggiunto

    return sorted_tests


# Ottieni i nomi dei test unici
ptrtname = ptr["TestName"].unique() if not ptr.is_empty() else []
ftrtname = ftr["TestName"].unique() if not ftr.is_empty() else []

# Ordina i nomi dei test
ptrtname = ordina_test_name(ptrtname)
ftrtname = ordina_test_name(ftrtname)

# Visualizza l'intestazione
display(Markdown("<p style='text-align:right;color:red'>ST Restricted</p>"))
display(Markdown("<a id='TableOfContent'></a>"))
display(Markdown("## Table of Contents"))
display(Markdown("---"))

content = ""

if len(ptrtname) != 0 and len(ftrtname) != 0:
    for tname in ptrtname:
        content = content + "<a class='btn' href='#" + tname + "'> " + tname + " </a>\n"
    for tname in ftrtname:
        content = content + "<a class='btn' href='#" + tname + "'> " + tname + " </a>\n"
elif len(ptrtname) != 0:
    for tname in ptrtname:
        content = content + "<a class='btn' href='#" + tname + "'> " + tname + " </a>\n"
elif len(ftrtname) != 0:
    for tname in ftrtname:
        content = content + "<a class='btn' href='#" + tname + "'> " + tname + " </a>\n"

# Visualizza il contenuto del sommario
display(
    Markdown(
        "<div class='contentconteiner'>\n"
        + content
        + "<a class='btn' href='#conclusions'>Conclusions</a></div>"
    )
)
ptr = ptr.to_pandas()
ftr = ftr.to_pandas()
del content
```

```python
results_df = pd.DataFrame()

# Supponendo che ftr e ftrtname siano già definiti
if not ftr.empty:
    uty.write_log("Start FTR data analysis")
    
    # DataFrame per memorizzare i risultati aggregati
    results_df = pd.DataFrame(columns=["TestName", "Pass", "Fail", "Yield"])
    
    for tname in ftrtname:
        uty.write_log(tname)
        display(
            Markdown(
                "<a id='"
                + tname
                + "'></a><p style='text-align:right;color:red'>ST Restricted</p>"
            )
        )
        display(Markdown("### " + tname))
        display(Markdown("---"))
        td = pd.DataFrame(ftr[(ftr["TestName"] == tname)])
        td.rename(
            columns={
                "COM": "Composite",
                "WAFER": "Wafer",
                "X_COORD": "XId",
                "Y_COORD": "YId",
            },
            inplace=True,
        )
        if td.empty:
            continue
        if pd.unique(td.pltype)[0] == "BPLTEMP":
            if {"XId", "YId", "RESULT"}.issubset(td.columns):
                colorscale = [[0, "red"], [1, "green"]]
                # Crea la heatmap
                heatmap = go.Heatmap(
                    x=td["XId"],
                    y=td["YId"],
                    z=td["RESULT"],
                    colorscale=colorscale,
                    showscale=False,
                    zmin=0,  # Valore minimo della scala
                    zmax=1,  # Valore massimo della scala
                )
                ## Crea la figura
                fig = go.Figure(data=[heatmap])

                # Imposta lo sfondo bianco
                fig.update_layout(plot_bgcolor="white")

                # Rimuove i tick dalle assi
                fig.update_xaxes(showticklabels=False)
                fig.update_yaxes(showticklabels=False)

                # Rimuove la scala colore
                heatmap.coloraxis = None

                # Mostra la figura
                fig.show()
            
            # Ensure every TestName has both PASS and FAIL entries
            all_tests = td["TestName"].unique()
            all_results = pd.DataFrame([(test, result) for test in all_tests for result in [0, 1]], columns=["TestName", "RESULT"])

            # Merge with the original data to fill missing combinations with zeros
            td_full = all_results.merge(td.groupby(["TestName", "RESULT"])["RESULT"].agg(
                TOTAL_RESULT="count"
            ).reset_index(), on=["TestName", "RESULT"], how="left").fillna(0).astype({"TOTAL_RESULT": int})

            # Calculate percentage
            td_full["PERCENT"] = td_full["TOTAL_RESULT"] / td_full.groupby("TestName")["TOTAL_RESULT"].transform("sum") * 100
            td_full["PERCENTstr"] = td_full["PERCENT"].round(2).astype(str) + "%"

            # Trace for PASS
            trace_pass = go.Bar(
                x=td_full[td_full["RESULT"] == 1]["PERCENT"],
                y=td_full[td_full["RESULT"] == 1]["TestName"],
                orientation="h",
                name="",
                text=td_full[td_full["RESULT"] == 1]["PERCENTstr"] + " (" + td_full[td_full["RESULT"] == 1]["TOTAL_RESULT"].astype(str) + " duts)",
                hovertemplate="Percent: %{customdata[0]}<br>"
                            + "Counter: %{customdata[1]}<br>"
                            + "SBIN: %{customdata[2]}",
                customdata=td_full[td_full["RESULT"] == 1][["PERCENTstr", "TOTAL_RESULT", "TestName"]].values,
                marker=dict(color=STblue),
            )

            # Trace for FAIL
            trace_fail = go.Bar(
                x=td_full[td_full["RESULT"] == 0]["PERCENT"],
                y=td_full[td_full["RESULT"] == 0]["TestName"],
                orientation="h",
                name="",
                text=td_full[td_full["RESULT"] == 0]["PERCENTstr"] + " (" + td_full[td_full["RESULT"] == 0]["TOTAL_RESULT"].astype(str) + " duts)",
                hovertemplate="Percent: %{customdata[0]}<br>"
                            + "Counter: %{customdata[1]}<br>"
                            + "SBIN: %{customdata[2]}",
                customdata=td_full[td_full["RESULT"] == 0][["PERCENTstr", "TOTAL_RESULT", "TestName"]].values,
                marker=dict(color=STred),
            )

            # Layout for the chart
            layout = go.Layout(
                xaxis=dict(
                    title="",
                    showgrid=True,
                    gridcolor="lightgray",
                    zeroline=False,
                    showticklabels=True,
                ),
                plot_bgcolor="white",
                hoverlabel_bgcolor=STcyan,
                height=400,
                showlegend=False,
            )

            # Create the figure
            fig = go.Figure(data=[trace_fail, trace_pass], layout=layout)
            fig.update_traces(hoverlabel=dict(
                font=dict(color=STblue),
                bordercolor=STblue,
                bgcolor=STcyan
            ))
            fig.update_traces(hovertemplate=None)
            fig.show()

                
        if pd.unique(td.pltype)[0] == "BPLVDD":
            # crea un dataframe con i conteggi di 1 per ogni valore di VDD, temperatura e Corner
            df = (
                td.groupby(["Volt", "°C", "Corner"])["RESULT"]
                .agg(lambda x: sum(x == 1))
                .reset_index()
            )

            STtemplate = go.layout.Template()
            STtemplate.layout = go.Layout(
                plot_bgcolor="white",
                xaxis=dict(gridcolor="#ebf0f8", zerolinecolor="#dee3ea"),
                yaxis=dict(gridcolor="#ebf0f8"),
            )

            # crea il grafico
            fig = px.line(
                df,
                x="Volt",
                y="RESULT",
                color="°C",
                facet_row="Corner",
                color_discrete_sequence=STPalette,
                template=STtemplate,
            )
            fig.update_traces(mode="markers+lines", hovertemplate=None)
            fig.update_layout(
                height=400 * (td["Corner"].nunique() + 1),
                # plot_bgcolor="white",
                # paper_bgcolor="white",
                # template="plotly_white",
                hovermode="x unified",
            )
            fig.update_yaxes(title_text="Count")
            fig.show() 

            pv = td.groupby(["°C", "Corner", "Volt"], as_index=False).agg(
                Pass=("RESULT", lambda x: (x == 1).sum()),
                Fail=("RESULT", lambda x: (x == 0).sum()),
                Gross=("RESULT", "size"),
                Yield=("RESULT", lambda x: f"{round((x == 1).sum() / len(x) * 100, 2)} %"),
            )

            pv = pd.melt(
                pv,
                id_vars=["°C", "Corner", "Volt"],
                value_vars=["Pass", "Fail", "Gross", "Yield"],
                var_name="Metric",
                value_name="Value",
            )
            pv = pv.pivot(
                index=["°C", "Corner", "Metric"], columns="Volt", values="Value"
            ).reset_index()
            pv["Code"] = pv["Metric"].apply(
                lambda x: 1
                if x == "Pass"
                else (2 if x == "Fail" else (3 if x == "Gross" else 4))
            )
            pv = (
                pv.sort_values(by=["°C", "Corner", "Code"], ascending=[True, True, True])
                .drop("Code", axis=1)
                .set_index(["°C", "Corner", "Metric"])
            )
            pd.options.display.max_rows = None
            display(pv)
        
        # Aggiungi i risultati aggregati al DataFrame results_df
        pass_count = td[td["RESULT"] == 1].shape[0]
        fail_count = td[td["RESULT"] == 0].shape[0]
        percent_pass = (pass_count / (pass_count + fail_count)) * 100 if (pass_count + fail_count) > 0 else 0
        
        new_row = pd.DataFrame({
            "TestName": [tname],
            "Pass": [pass_count],
            "Fail": [fail_count],
            "Yield": [str(round(percent_pass, 2))+"%"]
        })
        
        results_df = results_df.dropna(axis=1, how='all')
        new_row = new_row.dropna(axis=1, how='all')
        results_df = pd.concat([results_df, new_row], ignore_index=True)
        del new_row, pass_count, percent_pass, fail_count
    uty.write_log("FTR analysis done")
```

```python
uty.write_log("Write conclusion")

# Lists to store tests that pass, fail, and not binned
pass_tests, fail_tests, not_binned_tests = [], [], []
pass_tests_results_df, fail_tests_results_df = [], []

if not statistics.empty:
    # Filter columns in statistics
    statistics = statistics[[
        "TestName",
        "Tested Part",
        "Low Limit",
        "Min",
        "Mean",
        "Max",
        "High Limit",
        "Std",
        "Unit",
        "Cp",
        "CpK",
        "Kurtosis",
        "Fail Duts",
        "Yield",
    ]]
    
    for _, row in statistics.iterrows():
        if row["Low Limit"] == "-" or row["High Limit"] == "-":
            not_binned_tests.append(row)
        elif row["Cp"] != "-" and row["Cp"] < 10:
            fail_tests.append(row)
        elif row["Yield"] != "100.00%" :
            fail_tests.append(row)
        else:
            pass_tests.append(row)


# Check tests in results_df
for _, row in results_df.iterrows():
    percent_pass_value = float(row["Yield"].strip('%'))
    (fail_tests_results_df if percent_pass_value < 100 else pass_tests_results_df).append(row)

# Determine the overall conclusion
all_tests_within_spec = not fail_tests and not fail_tests_results_df
conclusion_html = "<h1 id='conclusions'>Conclusions</h1>"
conclusion_html += "<h3>{}</h3>".format(
    "All tests are within specifications."
    if all_tests_within_spec
    else f"{len(fail_tests) + len(fail_tests_results_df)} out of {statistics.shape[0] + results_df.shape[0]} tests are not within specifications."
)

def apply_style(row):
    if row["Yield"] != "100.00%":
        return ['{}'.format(uty.color_yield(val))if col == 'Yield' else '' for col, val in row.items()]
    else:
        return ['{}'.format(uty.color_cp(val)) if col == 'Cp' else '' for col, val in row.items()]

# Tests that fail
if fail_tests or fail_tests_results_df:
    conclusion_html += "<h2>Fail Tests:</h2>"
    if fail_tests:
        conclusion_html += "<h3>Parametric Tests statistics</h3>"
        conclusion_html += (
            pd.DataFrame(fail_tests)
            .set_index("TestName")
            .style.apply(apply_style, axis=1)
            .format(custom_format)
            .to_html(index=False, escape=False)
        )
    if fail_tests_results_df:
        conclusion_html += "<h3>Functional Tests results</h3>"
        conclusion_html += results_df[results_df["Yield"].apply(lambda x: float(x.strip('%')) < 100)].to_html(index=False)

# Tests that pass
conclusion_html += "<h2>Pass Tests:</h2>"
if pass_tests:
    conclusion_html += "<h3>Parametric Tests statistics</h3>"
    conclusion_html += (
        pd.DataFrame(pass_tests)
        .set_index("TestName")
        .style.map(uty.color_cp, subset=["Cp"])
        .format(custom_format)
        .to_html(index=True, escape=False)
    )
if pass_tests_results_df:
    conclusion_html += "<h3>Functional Tests results</h3>"
    conclusion_html += results_df[results_df["Yield"].apply(lambda x: float(x.strip('%')) >= 100)].set_index("TestName").style.to_html(index=False)

# Tests that are not binned
if not_binned_tests:
    conclusion_html += "<h2>Not Binned Tests:</h2>"
    conclusion_html += "<h3>Parametric Tests statistics</h3>"
    conclusion_html += (
        pd.DataFrame(not_binned_tests)
        .set_index("TestName")
        .style.map(uty.color_cp, subset=["Cp"])
        .format(custom_format)
        .to_html(index=False, escape=False)
    )

# Display conclusions
display(HTML(conclusion_html))
```
